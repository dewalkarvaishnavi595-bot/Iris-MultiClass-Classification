```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

```python
print(df.isnull().sum())
```

```python
print(df["species"].value_counts())
```

```python
sns.pairplot(df, hue="species")
plt.show()
```

```python
X = df.drop("species", axis=1)
y = df["species"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=1000))
])

print("Model created successfully")
```

```python
logistic_model.fit(X_train, y_train)

y_pred_lr = logistic_model.predict(X_test)

print(y_pred_lr)
```

```python
accuracy_lr = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Accuracy:", accuracy_lr)
```

```python
precision_lr = precision_score(
    y_test,
    y_pred_lr,
    average="weighted"
)

recall_lr = recall_score(
    y_test,
    y_pred_lr,
    average="weighted"
)

f1_lr = f1_score(
    y_test,
    y_pred_lr,
    average="weighted"
)

print("Precision:", precision_lr)
print("Recall:", recall_lr)
print("F1-Score:", f1_lr)
```

```python
print(classification_report(y_test, y_pred_lr))
```

```python
cm_lr = confusion_matrix(y_test, y_pred_lr)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d",
    xticklabels=logistic_model.classes_,
    yticklabels=logistic_model.classes_
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression Confusion Matrix")

plt.show()
```

```python
y_pred_dt = tree_model.predict(X_test)

print(y_pred_dt)
```

```python
accuracy_dt = accuracy_score(y_test, y_pred_dt)

precision_dt = precision_score(
    y_test,
    y_pred_dt,
    average="weighted"
)

recall_dt = recall_score(
    y_test,
    y_pred_dt,
    average="weighted"
)

f1_dt = f1_score(
    y_test,
    y_pred_dt,
    average="weighted"
)

print("Decision Tree")
print("--------------------")
print("Accuracy :", accuracy_dt)
print("Precision:", precision_dt)
print("Recall   :", recall_dt)
print("F1-Score :", f1_dt)
```

```python
print(classification_report(y_test, y_pred_dt))
```

```python
cm_dt = confusion_matrix(y_test, y_pred_dt)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_dt,
    annot=True,
    fmt="d",
    xticklabels=tree_model.classes_,
    yticklabels=tree_model.classes_
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Decision Tree Confusion Matrix")

plt.show()
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree"
    ],
    "Accuracy": [
        accuracy_lr,
        accuracy_dt
    ],
    "Precision": [
        precision_lr,
        precision_dt
    ],
    "Recall": [
        recall_lr,
        recall_dt
    ],
    "F1-Score": [
        f1_lr,
        f1_dt
    ]
})

results.round(4)
```

```python
X_2d = df[["petal_length", "petal_width"]]
y_2d = df["species"]
```

```python
X_train_2d, X_test_2d, y_train_2d, y_test_2d = train_test_split(
    X_2d,
    y_2d,
    test_size=0.20,
    random_state=42,
    stratify=y_2d
)
```

```python
model_2d = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=1000))
])

model_2d.fit(X_train_2d, y_train_2d)
```

```python
x_min = X_2d["petal_length"].min() - 0.5
x_max = X_2d["petal_length"].max() + 0.5

y_min = X_2d["petal_width"].min() - 0.5
y_max = X_2d["petal_width"].max() + 0.5

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.02),
    np.arange(y_min, y_max, 0.02)
)

grid = pd.DataFrame({
    "petal_length": xx.ravel(),
    "petal_width": yy.ravel()
})

Z = model_2d.predict(grid)

class_mapping = {
    "setosa": 0,
    "versicolor": 1,
    "virginica": 2
}

Z = pd.Series(Z).map(class_mapping).values
Z = Z.reshape(xx.shape)
```

```python
plt.figure(figsize=(9, 6))

plt.contourf(xx, yy, Z, alpha=0.3)

sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species",
    s=70
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Logistic Regression Decision Boundary")

plt.show()
```

```python
from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc
from sklearn.multiclass import OneVsRestClassifier

# Convert classes into binary format
classes = sorted(y.unique())
y_test_binary = label_binarize(y_test, classes=classes)

# Train Logistic Regression with One-vs-Rest
roc_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", OneVsRestClassifier(
        LogisticRegression(max_iter=1000)
    ))
])

roc_model.fit(X_train, y_train)

# Get prediction probabilities
y_score = roc_model.predict_proba(X_test)

# Plot ROC curves
plt.figure(figsize=(8, 6))

for i, class_name in enumerate(classes):
    fpr, tpr, _ = roc_curve(
        y_test_binary[:, i],
        y_score[:, i]
    )

    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f"{class_name} (AUC = {roc_auc:.2f})"
    )

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Multiclass ROC Curve - Logistic Regression")
plt.legend()
plt.show()
```
